```python
def calculator(number1, number2):

    def add():
        return number1 + number2

    def subtract():
        return number1 - number2

    print("Addition:", add())
    print("Subtraction:", subtract())


calculator(20, 10)
```

```python
def register_user(username, password):

    def validate_username():
        return len(username) >= 5

    def validate_password():
        return len(password) >= 8

    if not validate_username():
        return "Username must contain at least 5 characters"

    if not validate_password():
        return "Password must contain at least 8 characters"

    return "Registration successful"


print(register_user("bharath", "python123"))
print(register_user("ram", "1234"))
```

```python
def outer_function(name):
    greeting = "Hello"

    def middle_function():
        message = f"{greeting}, {name}"

        def inner_function():
            print(message)
            print("Welcome to Python programming!")

        inner_function()

    middle_function()

outer_function("Edukron")
```

```python
def calculate(number1, number2):

    def add_numbers():
        total = number1 + number2

        def calculate_square():
            square = total ** 2
            return square

        return calculate_square()

    return add_numbers()


result = calculate(10, 5)
print("Square of the sum:", result)
```

```python
def shopping_bill(product, price, quantity):
    def calculate_total():
        total = price * quantity

        def apply_discount():
            if total >= 1000:
                discount = total * 0.10
            else:
                discount = 0
            final_amount = total - discount
            print("Product:", product)
            print("Original amount:", total)
            print("Discount:", discount)
            print("Final amount:", final_amount)

        apply_discount()

    calculate_total()
shopping_bill("Headphones", 400, 2)
```

```python
def factorial(n):
  if n==0 or n==1:
    return 1
  return n*factorial(n-1)
factorial(5)
```
