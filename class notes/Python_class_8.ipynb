```python
x=10
### This loop will run as long
### as the condition satisfy (True)
### Infinite loop
### Increment or decrement operator
x=10
while x>5:
  print("the value before decrement",x)
  x=x-1
  print("the value after decrement",x)
  print(" ")
```

```python
x=10
while x<20:
  x=x+1
  print(x,end=',')
```

```python
x=5
while x>=1:
  print(x*"*")
  x=x-1
```

```python
x=1
space=4
while x<=5:
  print(" "*space+x*"*")
  x=x+1
  space=space-1
```

```python
for i in range(1,10):

  if i==5:
    continue
  print(i,end=",")
```

```python
for i in "datascience":

  if i=="a":
    continue
  print(i,end=",")
```

```python
for i in range(1,6):
  for j in range(1,6):
    if j==3:
      break
    print([i,j],end=",")
```

```python
for i in range(1,6):
  for j in range(1,6):
    print([i,j],end=",")
    if j==3:
      continue
```

```python
for i in "data":
  for j in "da":
    if j=="a":
      break
    for k in "sci":
      print([i,j,k],end=',')
```
