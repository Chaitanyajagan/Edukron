Part 1: Dataset Creation

```python
import numpy as np

# Store names
stores = np.array([
    "Bengaluru",
    "Hyderabad",
    "Chennai",
    "Mumbai",
    "Pune"
])

# Product categories
categories = np.array([
    "Electronics",
    "Groceries",
    "Clothing",
    "Furniture",
    "Beauty",
    "Sports"
])

# Sales revenue in rupees
# Rows represent stores
# Columns represent product categories

sales_data = np.array([
    [850000, 420000, 310000, 275000, 190000, 225000],
    [780000, 460000, 295000, 310000, 210000, 240000],
    [920000, 390000, 340000, 260000, 185000, 215000],
    [1050000, 510000, 375000, 345000, 250000, 290000],
    [690000, 405000, 280000, 225000, 175000, 205000]
])

# Cost incurred in rupees
# Rows represent stores
# Columns represent product categories

cost_data = np.array([
    [640000, 315000, 220000, 205000, 125000, 160000],
    [590000, 350000, 210000, 225000, 140000, 170000],
    [700000, 295000, 245000, 195000, 120000, 155000],
    [790000, 390000, 265000, 250000, 165000, 205000],
    [530000, 310000, 205000, 170000, 115000, 145000]
])

print("Stores:")
print(stores)

print("\nProduct Categories:")
print(categories)

print("\nSales Data:")
print(sales_data)

print("\nCost Data:")
print(cost_data)
```

```python
'''

Question 1: Array Dimensions
The finance manager wants to confirm how many stores and product categories are present in the data.

Write NumPy code to display:

Number of dimensions in sales_data
Shape of sales_data
Total number of values
Data type of the array
Memory occupied by the array

'''

num_dimensions = sales_data.ndim
print("Number of dimension in sales_data: ",num_dimensions)

shape = sales_data.shape
print("Shape of sales_data: ",shape)

total_values = sales_data.size
print("Total number of values: ",total_values)

data_type = sales_data.dtype
print("Data type of the array: ",data_type)

memory_occupied = sales_data.nbytes
print("Memory occupied by the array: ",memory_occupied," bytes")
```

```python
'''

Question 2: Store and Category Count
Calculate the following:

Total number of stores
Total number of product categories
Total number of store-category combinations

'''

total_stores = stores.size
print("Total number of stores: ", total_stores)

total_categories = categories.size
print("Total number of product categories: ", total_categories)

total_combinations = total_stores * total_categories
print("Total number of store-category combinations: ", total_combinations)
```

Part 7: Profit Margin Analysis

```python
'''

Question 35: Profit Margin Matrix
Calculate the profit margin percentage for every store-category combination.

Use:
profit_margin = (profit_data / sales_data) * 100

Round the result to two decimal places.

'''

profit_margin = (profit_data / sales_data) * 100
profit_margin_rounded = np.round(profit_margin, 2)
print("Profit Margin Matrix (rounded to 2 decimal places):")
print(profit_margin_rounded)
```

```python
'''

Question 36: Store-Wise Profit Margin
Calculate the overall profit margin percentage for each store using:

Store Profit Margin =
Total Store Profit / Total Store Sales × 100

'''

store_profit_margins = (np.sum(profit_data, axis=1) / np.sum(sales_data, axis=1)) * 100
print("Store-Wise Profit Margins:")
print(store_profit_margins)
```

```python
'''

Question 37: Category-Wise Profit Margin
Calculate the overall profit margin percentage for each product category.

'''

category_profit_margins = (np.sum(profit_data, axis=0) / np.sum(sales_data, axis=0)) * 100
print("Category-Wise Profit Margins:")
print(category_profit_margins)
```

```python
'''

Question 38: Best Store by Profit Margin
Identify the store with the highest overall profit margin.

'''

best_store_index = np.argmax(store_profit_margins)
best_store_name = stores[best_store_index]
print(f"Best Store by Profit Margin: {best_store_name}")
```

```python
'''

Question 39: Best Category by Profit Margin
Identify the product category with the highest profit margin.

'''

best_category_index = np.argmax(category_profit_margins)
best_category_name = categories[best_category_index]
print(f"Best Category by Profit Margin: {best_category_name}")
```

```python
'''

Question 40: Low-Margin Transactions
Management considers a profit margin below 22% to be weak.

Find all store-category combinations where the profit margin is below 22%.

Display:

Store
Category
Sales
Cost
Profit
Profit margin
Use Boolean indexing.

'''

low_margin_indices = np.where(profit_margin < 22)
print("Low-Margin Transactions (Profit Margin < 22%):")
for store_idx, category_idx in zip(*low_margin_indices):
    store_name = stores[store_idx]
    category_name = categories[category_idx]
    sales_value = sales_data[store_idx, category_idx]
    cost_value = cost_data[store_idx, category_idx]
    profit_value = profit_data[store_idx, category_idx]
    profit_margin_value = profit_margin_rounded[store_idx, category_idx]

    print(f"Store: {store_name}, Category: {category_name}, Sales: {sales_value}, Cost: {cost_value}, Profit: {profit_value}, Profit Margin: {profit_margin_value}%")

```

Part 9: Target-Based Analysis

```python
sales_targets = np.array([
    800000,  # Electronics
    450000,  # Groceries
    320000,  # Clothing
    300000,  # Furniture
    200000,  # Beauty
    230000   # Sports
])
```

```python
'''

Question 48: Target Comparison
Compare every store’s category sales against sales_targets.

Use NumPy broadcasting.

'''

target_comparison = sales_data >= sales_targets
for store_idx in range(len(stores)):
    for category_idx in range(len(categories)):
        store_name = stores[store_idx]
        category_name = categories[category_idx]
        meets_target = target_comparison[store_idx, category_idx]
        print(f"Store: {store_name}, Category: {category_name}, Meets Target: {meets_target}")
```

```python
'''

Question 49: Target Achievement Matrix
Create a Boolean array where:

True means the store achieved the category target
False means the store failed to achieve the target

'''

target_achievement_matrix = sales_data >= sales_targets
print("Target Achievement Matrix (True = Achieved, False = Not Achieved):")
print(target_achievement_matrix)
```

```python
'''

Question 50: Target Achievement Count
Calculate the number of categories in which each store achieved its target.

'''
achievement_counts = np.sum(target_achievement_matrix, axis=1)
print("Target Achievement Counts:")
for store_idx, count in enumerate(achievement_counts):
    print(f"Store: {stores[store_idx]}, Achieved Targets: {count}")
```

```python
'''

Question 51: Best Target-Performing Store
Find the store that achieved the highest number of category targets.

'''

best_target_store_index = np.argmax(achievement_counts)
best_target_store_name = stores[best_target_store_index]
print(f"Best Target-Performing Store: {best_target_store_name}")
```

```python
'''

Question 52: Target Achievement by Category
Calculate how many stores achieved the target for each product category.

'''

target_achievement_by_category = np.sum(target_achievement_matrix, axis=0)
print("Target Achievement by Category:")
for category_idx, count in enumerate(target_achievement_by_category):
    print(f"Category: {categories[category_idx]}, Stores Achieved Target: {count}")
```

```python
'''

Question 53: Sales Target Variance
Calculate the difference between actual sales and target sales for every store.

Use:

Variance = Actual Sales - Target Sales

'''

sales_target_variance = sales_data - sales_targets

print("Sales Target Variance (Actual - Target) for each store:")
for store_idx, store in enumerate(stores):
    print(f"{store}: {sales_target_variance[store_idx]}")
```

```python
'''

Question 54: Percentage Above or Below Target
Calculate the percentage variance from the target using:

Percentage Variance =
(Actual Sales - Target Sales) / Target Sales × 100

'''

percentage_target = (sales_target_variance) / sales_targets * 100
print("The percentage variance from the target using:")
for store_idx, store in enumerate(stores):
    print(f"{store}: {percentage_target[store_idx]}")
```
