```python
bank_dataset=pd.read_excel(r'C:\Users\Hp\Downloads\Regression\data\application_test.xlsx')
```

```python
def fill_null_values(df):
    for column in df.columns:
        if df[column].isna().sum() > 0:
            if df[column].dtype in ['int64', 'float64']:
                df[column] = df[column].fillna(df[column].median())
    
    return df

bank_dataset = fill_null_values(bank_dataset)
bank_dataset.isna().sum()
```

```python
# bank_dataset=bank_dataset[['AMT_CREDIT','AMT_ANNUITY','AMT_GOODS_PRICE','DAYS_BIRTH','DAYS_EMPLOYED','DAYS_REGISTRATION','DAYS_ID_PUBLISH','CNT_FAM_MEMBERS']]
```

```python
y=bank_dataset['AMT_CREDIT']
x=bank_dataset[['AMT_INCOME_TOTAL','AMT_ANNUITY','AMT_GOODS_PRICE','DAYS_BIRTH','DAYS_EMPLOYED','CNT_CHILDREN','CNT_FAM_MEMBERS','EXT_SOURCE_1','EXT_SOURCE_2','EXT_SOURCE_3']]
```

```python
from sklearn.model_selection import train_test_split
```

```python
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=111)
```

```python
from sklearn.linear_model import LinearRegression
lr=LinearRegression()
lr.fit(x_train,y_train)
predicted_values=lr.predict(x_test)
print(lr.coef_)
```

```python
np.round(lr.coef_).astype(int)
```

```python
from sklearn.metrics import r2_score
r2_score(y_test.values,lr.predict(x_test))
```

```python
abs(predicted_values-y_test).mean()
```

```python
from sklearn.linear_model import LinearRegression, Ridge, Lasso, ElasticNet
from sklearn.neighbors import KNeighborsRegressor
from sklearn.metrics import r2_score
```

```python
lr = LinearRegression()

lr.fit(x_train, y_train)

predicted_values = lr.predict(x_test)

print("Linear Regression Coefficients:")
print(lr.coef_)

print("Linear Regression R2 Score:")
print(r2_score(y_test.values, predicted_values))
```

```python
ridge = Ridge(alpha=1.0)

ridge.fit(x_train, y_train)

ridge_pred = ridge.predict(x_test)

print("\nRidge Regression Coefficients:")
print(ridge.coef_)

print("Ridge Regression R2 Score:")
print(r2_score(y_test.values, ridge_pred))
```

```python
lasso = Lasso(alpha=0.1)

lasso.fit(x_train, y_train)

lasso_pred = lasso.predict(x_test)

print("\nLasso Regression Coefficients:")
print(lasso.coef_)

print("Lasso Regression R2 Score:")
print(r2_score(y_test.values, lasso_pred))
```

```python
elastic = ElasticNet(alpha=0.1, l1_ratio=0.5)

elastic.fit(x_train, y_train)

elastic_pred = elastic.predict(x_test)

print("\nElasticNet Regression Coefficients:")
print(elastic.coef_)

print("ElasticNet Regression R2 Score:")
print(r2_score(y_test.values, elastic_pred))
```

```python
knn = KNeighborsRegressor(n_neighbors=5)

knn.fit(x_train, y_train)

knn_pred = knn.predict(x_test)

print("\nKNN Regression R2 Score:")
```

```python
import numpy as np
import pandas as pd
```

```python
from sklearn.metrics import (
    r2_score,
    mean_absolute_error,
    mean_squared_error,
    mean_absolute_percentage_error
)
```

```python
lr_r2 = r2_score(y_test, predicted_values)
lr_mae = mean_absolute_error(y_test, predicted_values)
lr_mse = mean_squared_error(y_test, predicted_values)
lr_rmse = np.sqrt(lr_mse)
lr_mape = mean_absolute_percentage_error(y_test, predicted_values) * 100
```

```python
ridge_r2 = r2_score(y_test, ridge_pred)
ridge_mae = mean_absolute_error(y_test, ridge_pred)
ridge_mse = mean_squared_error(y_test, ridge_pred)
ridge_rmse = np.sqrt(ridge_mse)
ridge_mape = mean_absolute_percentage_error(y_test, ridge_pred) * 100
```

```python
lasso_r2 = r2_score(y_test, lasso_pred)
lasso_mae = mean_absolute_error(y_test, lasso_pred)
lasso_mse = mean_squared_error(y_test, lasso_pred)
lasso_rmse = np.sqrt(lasso_mse)
lasso_mape = mean_absolute_percentage_error(y_test, lasso_pred) * 100
```

```python
elastic_r2 = r2_score(y_test, elastic_pred)
elastic_mae = mean_absolute_error(y_test, elastic_pred)
elastic_mse = mean_squared_error(y_test, elastic_pred)
elastic_rmse = np.sqrt(elastic_mse)
elastic_mape = mean_absolute_percentage_error(y_test, elastic_pred) * 100
```

```python
knn_r2 = r2_score(y_test, knn_pred)
knn_mae = mean_absolute_error(y_test, knn_pred)
knn_mse = mean_squared_error(y_test, knn_pred)
knn_rmse = np.sqrt(knn_mse)
knn_mape = mean_absolute_percentage_error(y_test, knn_pred) * 100
```

```python
comparison_table = pd.DataFrame({

    "Model": [
        "Linear Regression",
        "Ridge Regression",
        "Lasso Regression",
        "ElasticNet Regression",
        "KNN Regression"
    ],

    "R2 Score": [
        lr_r2,
        ridge_r2,
        lasso_r2,
        elastic_r2,
        knn_r2
    ],

    "MAE": [
        lr_mae,
        ridge_mae,
        lasso_mae,
        elastic_mae,
        knn_mae
    ],

    "MSE": [
        lr_mse,
        ridge_mse,
        lasso_mse,
        elastic_mse,
        knn_mse
    ],

    "RMSE": [
        lr_rmse,
        ridge_rmse,
        lasso_rmse,
        elastic_rmse,
        knn_rmse
    ],

    "MAPE (%)": [
        lr_mape,
        ridge_mape,
        lasso_mape,
        elastic_mape,
        knn_mape
    ]
})


# Round values
comparison_table = comparison_table.round(4)


# Sort by R2 Score
comparison_table = comparison_table.sort_values(
    by="R2 Score",
    ascending=False
)


# Print final comparison table
print(comparison_table.to_string(index=False))
```
