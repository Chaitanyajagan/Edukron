```python
'''

Assignment 16: Reverse a String
Write a function named reverse_string().

Requirements
Accept a string.
Return its reversed version.
Solve it first using slicing.
Write another function that reverses it using a loop.

'''

def reverse_string(k):
    return k[::-1]

def reverse_string_loop(text):
    reversed_text = ""
    for char in text:
        reversed_text = char + reversed_text
    return reversed_text

str = input("Enter text: ")
print("Reversed text:",reverse_string(str))
print("Reversing string using loop:",reverse_string_loop(str))
```

```python
'''

Assignment 17: Palindrome Checker
Write a function named is_palindrome().

Requirements
Accept a string.
Ignore spaces and letter case.
Return True when the string is a palindrome.
Print a meaningful result.

'''

def is_palindrome(text):
    ignore = "".join(text.split()).lower()
    if ignore == ignore[::-1]:
        print("The text is a palindrome.")
    else:
        print("The text is not a palindrome.")

text = input("Enter text: ")
is_palindrome(text)
```

```python
'''

Assignment 19: Find Maximum in a List
Write a function named find_list_maximum().

Requirements
Accept a list of numbers.
Do not use max().
Return the largest list element.
Handle an empty list by returning an appropriate message.

'''

def find_list_maximum(num):
    maxi = num[0]
    for i in num:
        if i > maxi:
            maxi = i
    return maxi

li = list(map(int, input("Enter numbers: ").split()))
print(li)
print("Maximum:", find_list_maximum(li))
```

```python
'''

Assignment 20: Count Positive, Negative and Zero Values
Write a function named count_number_types().

Requirements
Accept a list of integers.
Count positive numbers, negative numbers, and zeros.
Return all three counts.
Use tuple unpacking to receive the returned values.

'''

def count_number_types(x):
    count_positive = 0
    count_negative = 0
    count_zero = 0
    for i in x:
        if i > 0:
            count_positive += 1
        elif i < 0:
            count_negative += 1
        else:
            count_zero += 1
    return (count_positive, count_negative, count_zero)

li = list(map(int, input("Enter numbers: ").split()))
print(li)
x = count_number_types(li)
print("Positive numbers:",x[0])
print("Negative numbers:",x[1])
print("Zeros:",x[2])
```

```python
'''

Assignment 31: Return Multiple Values
Write a function named number_statistics().

Requirements
Accept a list of numbers.
Return minimum, maximum, total, and average.
Use tuple unpacking to store returned values.
You may use built-in functions in this assignment.

'''

def number_statistics(numbers):
  if not numbers:
    return None, None, 0, 0

  low = min(numbers)
  high = max(numbers)
  total = sum(numbers)
  avg = total / len(numbers)

  return (low, high, total, avg)


data = [10, 20, 30, 40, 50]
minimum, maximum, total, average = number_statistics(data)

print(f"Min: {minimum}")
print(f"Max: {maximum}")
print(f"Total: {total}")
print(f"Average: {average}")
```

```python
'''

Assignment 32: Function Returning a Dictionary
Write a function named analyze_text().

Requirements
Accept a string.
Return a dictionary containing:
    Total characters
    Total words
    Total vowels
    Total digits
    Total spaces
Display every result clearly.

'''

def analyze_text(text):
    vowels = 'aeiouAEIOU'
    a = {
        "total_char": len(text),
        "total_word": len(text.split()),
        "total_vowels": sum(1 for char in text if char in vowels),
        "total_digits": sum(1 for char in text if char.isdigit()),
        "total_spaces": sum(1 for char in text if char.isspace())
    }
    return a

input_text = input("Enter text: ")
analysis = analyze_text(input_text)
for key, value in analysis.items():
    print(f"{key.replace('_',' ').title()}: {value}")

    
```

```python
'''

Assignment 34: Prime Number Function
Write a function named is_prime().

Requirements
Accept one integer.
Return True when the number is prime.
Return False otherwise.
Use the function to print all prime numbers from 1 to 100.

'''

def is_prime(num):
    if num <= 1:
        return False

    for i in range(2, num):
        if num % i == 0:
            return False

    return True

number = int(input("Enter an integer: "))

if is_prime(number):
    print(number, "is a Prime Number.")
else:
    print(number, "is Not a Prime Number.")

print("\nPrime numbers from 1 to 100:")
for i in range(1, 101):
    if is_prime(i):
        print(i, end=" ")
```

```python
'''

Assignment 35: Fibonacci Function
Write a function named generate_fibonacci().

Requirements
Accept the number of terms.
Return the complete Fibonacci series as a list.
Handle zero and negative term counts.
Do not use recursion in this assignment.

'''

def generate_fibonacci(n):
    if n <= 0:
        return []
    if n == 1:
        return [0]

    series = [0, 1]
    for _ in range(2, n):
        series.append(series[-1] + series[-2])
    return series

terms = int(input("Enter number of Fibonacci terms: "))
print("Fibonacci series:", generate_fibonacci(terms)) 
```

```python
'''

Assignment 89: Closure-Based Multiplier
Write a function named create_multiplier().

Requirements
The outer function should accept a multiplier value.
Define an inner function that accepts a number.
The inner function should multiply the number by the outer multiplier.
Return the inner function.
Create separate double, triple, and ten-times functions.

'''

def create_multiplier(multiplier):
    def multiplier_function(number):
        return number * multiplier
    return multiplier_function

double = create_multiplier(2)
triple = create_multiplier(3)
ten_times = create_multiplier(10)
print("Double:", double(5))
print("Triple:", triple(5))
print("Ten times:", ten_times(5))
```

```python
'''

Assignment 90: Closure-Based Discount Calculator
Write a function named create_discount().

Requirements
The outer function should accept a discount percentage.
The inner function should accept a product price.
Return the final price after applying the discount.
Create separate functions for 10%, 20%, and 30% discounts.

'''

def create_discount(discount_percentage):
    def discount_function(price):
        return price - (price * discount_percentage / 100)
    return discount_function

ten_percent = create_discount(10)
twenty_percent = create_discount(20)
thirty_percent = create_discount(30)
print("Price after 10% discount:", ten_percent(100))
print("Price after 20% discount:", twenty_percent(100))
print("Price after 30% discount:", thirty_percent(100))
```

```python
'''

Assignment 91: Closure-Based Counter
Write a function named create_counter().

Requirements
Define a variable count = 0 in the outer function.
Define an inner function that increases the count.
Use nonlocal to modify the enclosing variable.
Return the inner function.
Call the returned function at least five times.

'''

def create_counter():
    count = 0
    def counter():
        nonlocal count
        count += 1
        return count
    return counter

counter = create_counter()
print("Counter:", counter())
print("Counter:", counter())
print("Counter:", counter())
print("Counter:", counter())
print("Counter:", counter())
```

```python
'''

Assignment 92: Closure-Based Bank Balance
Write a function named create_bank_account().

Requirements
Store the balance in the outer function.
Define inner functions for:
    Deposit
    Withdrawal
    Balance enquiry
Use nonlocal to update the balance.
Reject withdrawals greater than the available balance.
Return all three inner functions.

'''

def create_bank_account(initial_balance):
    balance = initial_balance

    def deposit(amount):
        nonlocal balance
        balance += amount
        return balance

    def withdraw(amount):
        nonlocal balance
        if amount > balance:
            return "Insufficient funds"
        balance -= amount
        return balance

    def get_balance():
        return balance

    return deposit, withdraw, get_balance

deposit, withdraw, get_balance = create_bank_account(1000)
print("Initial Balance:", get_balance())
print("After depositing 500:", deposit(500))
print("After withdrawing 300:", withdraw(300))
```

```python
'''

Assignment 96: Modify a Global Variable
Create a global variable named total_students.

Requirements
Initially assign a value of 100.
Define a function named add_students().
Use the global keyword to update the global value.
Ask the user how many students should be added.
Print the value before and after calling the function.

'''

total_students = 100

def add_students():
    global total_students
    num_students = int(input("How many students should be added? "))
    print(f"Adding {num_students} students.")
    total_students += num_students

print("Total students before adding:", total_students)
add_students()
print("Total students after adding:", total_students)
```

```python
'''

Assignment 98: Modify an Enclosing Variable Using nonlocal
Write a function named outer_counter().

Requirements
Define count = 10 in the outer function.
Define an inner function.
Use nonlocal count.
Increase the value by 5.
Print the value before and after calling the inner function.

'''

def outer_counter():
    count = 10
    print("Count before inner function:", count)
    
    def inner_function():
        nonlocal count
        count += 5
        print("Count inside inner function:", count)
    
    inner_function()
    print("Count after inner function:", count)
outer_counter()
```

```python
'''

Assignment 99: LEGB Name Resolution
Create variables with the same name at multiple scope levels.

Requirements
Use the variable name value in:

    Global scope
    Enclosing function scope
    Local function scope
Print the variable from different locations.

Tasks
Print the global value.
Print the enclosing value from the outer function.
Print the local value from the inner function.
Remove the local declaration and observe the result.
Remove the enclosing declaration and observe the result.
Explain the LEGB lookup order in comments.

LEGB Order
L → Local
E → Enclosing
G → Global
B → Built-in

'''

def demonstrate_LEGB():
    value = "Enclosing Value"
    
    def inner_function():
        value = "Local Value"
        print("Inside inner function:", value)
    
    inner_function()
    print("Inside outer function:", value)

demonstrate_LEGB()
value = "Global Value"
print("Global value:", value)
```

```python
'''

Assignment 100: Complete Functional Programming Project
Develop a complete Employee Performance Analysis System using functions, lambda expressions, map(), filter(), reduce(), nested functions, closures, and LEGB scope.

'''

company_name = "Edukron"

employees = [
    {"id": 101, "name": "Ravi", "department": "IT", "salary": 60000, "performance_score": 88, "experience": 5},
    {"id": 102, "name": "Anitha", "department": "HR", "salary": 45000, "performance_score": 72, "experience": 4},
    {"id": 103, "name": "Kiran", "department": "IT", "salary": 75000, "performance_score": 94, "experience": 7},
    {"id": 104, "name": "Meena", "department": "Finance", "salary": 55000, "performance_score": 81, "experience": 6},
    {"id": 105, "name": "Arjun", "department": "IT", "salary": 40000, "performance_score": 65, "experience": 2}
]

def display():
    for i in employees:
        print(i)

def names():
    print(list(map(lambda x: x["name"], employees)))

def high_performers():
    print(list(filter(lambda x: x["performance_score"] >= 80, employees)))

def department():
    dept = input("Enter department: ")
    data = list(filter(lambda x: x["department"].lower() == dept.lower(), employees))
    if data:
        print(data)
    else:
        print("Department not found")

def revised_salary():
    revised = list(map(lambda x: {
        "name": x["name"],
        "old_salary": x["salary"],
        "new_salary":
            x["salary"] * 1.15 if x["performance_score"] >= 90 else
            x["salary"] * 1.10 if x["performance_score"] >= 80 else
            x["salary"] * 1.05 if x["performance_score"] >= 70 else
            x["salary"]
    }, employees))

    for i in revised:
        print(i)

def total_salary():
    current = reduce(lambda a, b: a + b["salary"], employees, 0)

    revised = list(map(lambda x:
        x["salary"] * 1.15 if x["performance_score"] >= 90 else
        x["salary"] * 1.10 if x["performance_score"] >= 80 else
        x["salary"] * 1.05 if x["performance_score"] >= 70 else
        x["salary"], employees))

    new = reduce(lambda a, b: a + b, revised, 0)

    print("Current Salary :", current)
    print("Revised Salary :", new)
    print("Extra Expense  :", new - current)

def highest_paid():
    emp = reduce(lambda a, b: a if a["salary"] > b["salary"] else b, employees)
    print(emp)

def sort_employees():
    print("1.Name 2.Salary 3.Performance 4.Experience")
    ch = input("Enter choice: ")

    if ch == "1":
        data = sorted(employees, key=lambda x: x["name"])
    elif ch == "2":
        data = sorted(employees, key=lambda x: x["salary"])
    elif ch == "3":
        data = sorted(employees, key=lambda x: x["performance_score"])
    elif ch == "4":
        data = sorted(employees, key=lambda x: x["experience"])
    else:
        print("Invalid Choice")
        return

    for i in data:
        print(i)

def create_bonus_calculator(percent):
    def bonus(salary):
        return salary * percent / 100
    return bonus

def report():
    dept = input("Enter department: ")

    data = list(filter(lambda x: x["department"].lower() == dept.lower(), employees))

    if not data:
        print("Department not found")
        return

    report_title = "Department Report"

    print(report_title)
    print("Company:", company_name)
    print("Department:", dept)
    print("Employees:", len(data))
    print("Total Salary:", sum(map(lambda x: x["salary"], data)))
    print("Average Salary:", sum(map(lambda x: x["salary"], data)) / len(data))
    print("Average Performance:",
          sum(map(lambda x: x["performance_score"], data)) / len(data))

bonus5 = create_bonus_calculator(5)
bonus10 = create_bonus_calculator(10)
bonus15 = create_bonus_calculator(15)

while True:
    print("\n1.Display Employees")
    print("2.Employee Names")
    print("3.High Performers")
    print("4.Search Department")
    print("5.Revised Salaries")
    print("6.Total Salary")
    print("7.Highest Paid")
    print("8.Sort Employees")
    print("9.Department Report")
    print("10.Exit")

    choice = input("Enter choice: ")

    if choice == "1":
        display()
    elif choice == "2":
        names()
    elif choice == "3":
        high_performers()
    elif choice == "4":
        department()
    elif choice == "5":
        revised_salary()
    elif choice == "6":
        total_salary()
    elif choice == "7":
        highest_paid()
    elif choice == "8":
        sort_employees()
    elif choice == "9":
        report()
    elif choice == "10":
        print("Thank You")
        break
    else:
        print("Invalid Choice")
```
