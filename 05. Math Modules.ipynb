```python
# Create a scientific calculator that supports square root, factorial, logarithm, power, trigonometric functions, GCD, and LCM.

print("--- Scientific Calculator ---")
print("1. Square Root")
print("2. Factorial (x!)")
print("3. Logarithm (log(x))")
print("4. Power (x^y)")
print("5. Trigonometric Functions (sin, cos, tan)")
print("6. Greatest Common Divisor (GCD)")
print("7. Least Common Multiple (LCM)")
print("8. Exit")

while True:
    choice = input("\nSelect an operation (1-8): ")

    if choice == '8':
        print("Exiting calculator. Goodbye!")
        break

    if choice == '1':
        num = float(input("Enter number: "))
        if num < 0:
            print("Error: Cannot calculate the square root of a negative number.")
            break
        else:
            print(f"Square Root of {num}: {math.sqrt(num)}")
            break

    elif choice == '2':
        num = int(input("Enter a non-negative integer: "))
        if num < 0:
            print("Error: Factorial is not defined for negative numbers.")
            break
        else:
            print(f"Factorial of {num}: {math.factorial(num)}")
            break

    elif choice == '3':
        num = float(input("Enter number: "))
        base_input = input("Enter base (leave blank for natural log 'e'): ")
        if num <= 0:
            print("Error: Logarithm is only defined for positive numbers.")
            break
        elif base_input == '':
            print(f"Logarithm of {num}: {math.log(num)}")
            break
        else:
            print(f"Logarithm of {num} with base {base_input}: {math.log(num, float(base_input))}")
            break

    elif choice == '4':
        base = float(input("Enter base (x): "))
        exp = float(input("Enter exponent (y): "))
        print(f"Power of {base} raised to {exp}: {math.pow(base, exp)}")

    elif choice == '5':
        func = input("Choose function (sin, cos, tan): ").strip().lower()
        angle = float(input("Enter angle in degrees: "))
        radians = math.radians(angle) 

        if func == 'sin':
            print(f"sin({angle}°): {math.sin(radians)}")
        elif func == 'cos':
            print(f"cos({angle}°): {math.cos(radians)}")
        elif func == 'tan':
            print(f"tan({angle}°): {math.tan(radians)}")
        else:
            print("Invalid trigonometric function.")
            break

    elif choice == '6':
        a = int(input("Enter first integer: "))
        b = int(input("Enter second integer: "))
        print(f"GCD of {a} and {b}: {math.gcd(a, b)}")
        break

    elif choice == '7':
        a = int(input("Enter first integer: "))
        b = int(input("Enter second integer: "))
        
        if hasattr(math, 'lcm'):
            print(f"LCM of {a} and {b}: {math.lcm(a, b)}")
        else:
            lcm_result = abs(a * b) // math.gcd(a, b) if a and b else 0
            print(f"LCM of {a} and {b}: {lcm_result}")
        break

    else:
        print("Invalid choice. Please select a number from 1 to 8.")
```

```python
# Reverse a three-digit number using arithmetic operators.

number = int(input("Enter a three-digit number: ")) 
n = number
if 100 <= number <= 999:
    reversed_number = number % 10
    number //= 10
    reversed_number = reversed_number * 10 + number % 10
    number //= 10
    reversed_number = reversed_number * 10 + number
    print("Reversed number of", n, ":", reversed_number)
else:
    print("Please enter a valid three-digit number.")
```

```python
# Build a simple calculator using arithmetic and comparison operators.

print("--- Arithmetic & Comparison Calculator ---")
print("1. Addition (+)")
print("2. Subtraction (-)")
print("3. Multiplication (*)")
print("4. Division (/)")
print("5. Floor Division (//)")
print("6. Modulo (%)")
print("7. Exponentiation (**)")
print("8. Compare Two Numbers")
print("9. Exit")

while True:
    choice = input("\nSelect an operation (1-9): ")

    if choice == '9':
        print("Exiting calculator. Goodbye!")
        break

    if choice in ['1', '2', '3', '4', '5', '6', '7', '8']:
        num1 = int(input("Enter first number: "))
        num2 = int(input("Enter second number: "))

        # Arithmetic Operators
        if choice == '1':
            print(f"Result: {num1} + {num2} = {num1 + num2}")
            
        elif choice == '2':
            print(f"Result: {num1} - {num2} = {num1 - num2}")
            
        elif choice == '3':
            print(f"Result: {num1} * {num2} = {num1 * num2}")
            
        elif choice == '4':
            if num2 != 0:
                print(f"Result: {num1} / {num2} = {num1 / num2}")
            else:
                print("Error: Cannot divide by zero.")
                
        elif choice == '5':
            if num2 != 0:
                print(f"Result: {num1} // {num2} = {num1 // num2}")
            else:
                print("Error: Cannot divide by zero.")
                
        elif choice == '6':
            if num2 != 0:
                print(f"Result: {num1} % {num2} = {num1 % num2}")
            else:
                print("Error: Cannot divide by zero.")
                
        elif choice == '7':
            print(f"Result: {num1} ** {num2} = {num1 ** num2}")
            
        # Comparison Operators
        elif choice == '8':
            print(f"\n--- Comparing {num1} and {num2} ---")
            print(f"Equal to (==):        {num1 == num2}")
            print(f"Not equal to (!=):    {num1 != num2}")
            print(f"Greater than (>):     {num1 > num2}")
            print(f"Less than (<):        {num1 < num2}")
            print(f"Greater or equal (>=): {num1 >= num2}")
            print(f"Less or equal (<=):    {num1 <= num2}")
            
    else:
        print("Invalid choice. Please select a number from 1 to 9.")

    break
```

```python
# Create a grading system using comparison and logical operators.

print("--- Student Grading System ---")
score = int(input("Enter the student's score: "))

while True:

    if score < 0 or score > 100:
        print("Invalid input: Score must be between 0 and 100.")
        
    elif score >= 90 and score <= 100:
        print(f"Score: {score} -> Grade: A")
        
    elif score >= 80 and score < 90:
        print(f"Score: {score} -> Grade: B")
        
    elif score >= 70 and score < 80:
        print(f"Score: {score} -> Grade: C")
        
    elif score >= 60 and score < 70:
        print(f"Score: {score} -> Grade: D")
        
    elif score >= 0 and score < 60:
        print(f"Score: {score} -> Grade: F")

    break
```

```python
# Build a menu-driven calculator that uses arithmetic, comparison, logical, assignment, membership, identity, and bitwise operators wherever applicable.

print("--- Comprehensive Operator Calculator ---")
print("1. Arithmetic Operators")
print("2. Comparison Operators")
print("3. Logical Operators")
print("4. Assignment Operators")
print("5. Membership Operators")
print("6. Identity Operators")
print("7. Bitwise Operators")
print("8. Exit")

while True:
    choice = input("\nSelect an operator category to explore (1-8): ")

    if choice == '1':
        print("\n-- Arithmetic --")
        num1 = float(input("Enter first number: "))
        num2 = float(input("Enter second number: "))
        
        print(f"{num1} + {num2} = {num1 + num2}")
        print(f"{num1} - {num2} = {num1 - num2}")
        print(f"{num1} * {num2} = {num1 * num2}")
        if num2 != 0:
            print(f"{num1} / {num2} = {num1 / num2}")
            print(f"{num1} // {num2} (Floor Division) = {num1 // num2}")
            print(f"{num1} % {num2} (Modulo) = {num1 % num2}")
        print(f"{num1} ** {num2} (Exponentiation) = {num1 ** num2}")

    elif choice == '2':
        print("\n-- Comparison --")
        num1 = float(input("Enter first number: "))
        num2 = float(input("Enter second number: "))
        
        print(f"{num1} == {num2} is {num1 == num2}")
        print(f"{num1} != {num2} is {num1 != num2}")
        print(f"{num1} > {num2} is {num1 > num2}")
        print(f"{num1} <= {num2} is {num1 <= num2}")

    elif choice == '3':
        print("\n-- Logical (Truthiness) --")
        print("Note: In Python, 0 is False, and any other number is True.")
        x = float(input("Enter first number (x): "))
        y = float(input("Enter second number (y): "))
        
        print(f"{x} and {y} -> {x and y} (Returns first False, or last True)")
        print(f"{x} or {y} -> {x or y} (Returns first True, or last False)")
        print(f"not {x} -> {not x}")

    elif choice == '4':
        print("\n-- Assignment --")
        val = float(input("Enter a starting value: "))
        step = float(input("Enter a step value to apply: "))
        
        print(f"Original value: x = {val}")
        val += step
        print(f"After x += {step} -> x = {val}")
        val -= step
        print(f"After x -= {step} -> x = {val}")
        val *= step
        print(f"After x *= {step} -> x = {val}")

    elif choice == '5':
        print("\n-- Membership --")
        sequence = input("Enter a sentence or string of text: ")
        search = input("Enter a letter or word to search for: ")
        
        print(f"'{search}' in text -> {search in sequence}")
        print(f"'{search}' not in text -> {search not in sequence}")

    elif choice == '6':
        print("\n-- Identity --")
        print("Identity checks if two variables point to the exact same object in memory, not just if their values are equal.")
        
        list_a = [1, 2, 3]
        list_b = [1, 2, 3]
        list_c = list_a  # list_c points to the same memory location as list_a
        
        print("\nCreated variables:")
        print("list_a = [1, 2, 3]")
        print("list_b = [1, 2, 3] (Different object, same values)")
        print("list_c = list_a    (Same object as list_a)")
        
        print(f"\nlist_a == list_b (Values equal?): {list_a == list_b}")
        print(f"list_a is list_b (Same object?): {list_a is list_b}")
        print(f"list_a is list_c (Same object?): {list_a is list_c}")
        print(f"list_a is not list_b: {list_a is not list_b}")

    elif choice == '7':
        print("\n-- Bitwise --")
        print("Note: Bitwise operations require integers.")
        a = int(input("Enter first integer: "))
        b = int(input("Enter second integer: "))
        
        print(f"{a} & {b} (AND) -> {a & b}")
        print(f"{a} | {b} (OR) -> {a | b}")
        print(f"{a} ^ {b} (XOR) -> {a ^ b}")
        print(f"~{a} (NOT/Invert) -> {~a}")
        print(f"{a} << 1 (Left Shift by 1) -> {a << 1}")
        print(f"{a} >> 1 (Right Shift by 1) -> {a >> 1}")

    elif choice == '8':
            print("Exiting calculator. Goodbye!")
            break

    else:
        print("Invalid choice. Please select a number from 1 to 8.")

    break
```
