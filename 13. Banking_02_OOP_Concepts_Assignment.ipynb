### Q8. Protect a amount with a property for banking case 1; use sample reference `ACC-1008` and explain the result.

**Hints**

1. List the objects, their responsibilities, and their invariants before coding.
2. Choose composition for has-a relationships; use inheritance only for genuine is-a substitution.
3. Keep balances private and expose controlled deposit/withdraw/transfer behavior.

**Edge cases to test:** empty input, invalid types, boundary values, and a realistic failure path.

```python
class ProtectedBalanceAccount:
    def __init__(self, acc: str, initial: float):
        self.acc = acc
        self._balance = max(0.0, initial)

    @property
    def balance(self) -> float:
        return self._balance

    @balance.setter
    def balance(self, value: float):
        if value < 0:
            raise ValueError(f"Balance cannot be negative: {value}")
        self._balance = value

acc = ProtectedBalanceAccount("ACC-1008", 1000.0)
acc.balance = 1250.0
print(f"[{acc.acc}] Validated balance: ${acc.balance:.2f}")
```

### Q31. Inherit perishableaccount from account for banking case 1; use sample reference `ACC-1031` and explain the result.

**Hints**

1. List the objects, their responsibilities, and their invariants before coding.
2. Choose composition for has-a relationships; use inheritance only for genuine is-a substitution.
3. Keep balances private and expose controlled deposit/withdraw/transfer behavior.

**Edge cases to test:** empty input, invalid types, boundary values, and a realistic failure path.

```python
class BaseAccount:
    def __init__(self, acc: str, balance: float):
        self.acc, self.balance = acc, max(0.0, balance)

class TermDepositAccount(BaseAccount):
    def __init__(self, acc: str, balance: float, term_days: int):
        super().__init__(acc, balance)
        self.term_days = term_days

    def is_matured(self, days_held: int) -> bool:
        return days_held >= self.term_days

t_acc = TermDepositAccount("ACC-1031", 10000.0, 365)
print(f"[{t_acc.acc}] Term Deposit: ${t_acc.balance:.2f} | Matured (400d): {t_acc.is_matured(400)}")
```
