### Q1. Find a account code for banking case 1; use sample reference `ACC-1001` and explain the result.

**Hints**

### Q31. Parse a receipt line into named groups for banking case 1; use sample reference `ACC-1031` and explain the result.

**Hints**

1. Write representative valid and invalid samples before writing the pattern.
2. Use raw strings, named groups for extraction, and `fullmatch` for whole-field validation.
3. Never treat regex alone as authorization or full semantic validation; verify dates and checksums separately.

**Edge cases to test:** empty input, invalid types, boundary values, and a realistic failure path.

```python
import re

def parse_receipt_line(line):
    pattern = r'^(?P<date>\d{4}-\d{2}-\d{2})\s+(?P<ref>ACC-\d{4})\s+(?P<desc>.+?)\s+\$(?P<amount>\d+(?:\.\d{2})?)$'
    m = re.fullmatch(pattern, (line or '').strip())
    return m.groupdict() if m else None

entry = "2026-05-18 ACC-1031 International Wire Surcharge $45.50"
parsed = parse_receipt_line(entry)
print(f"[ACC-1031] Parsed Receipt: {parsed}")
```

### Q32. Validate an account reference with optional category prefix for banking case 1; use sample reference `ACC-1032` and explain the result.

**Hints**

3. Never treat regex alone as authorization or full semantic validation; verify dates and checksums separately.

**Edge cases to test:** empty input, invalid types, boundary values, and a realistic failure path.

```python
import re

def validate_categorized_account(ref_str):
    m = re.fullmatch(r'^(?:(?P<category>[A-Z]{2,5})-)?(?P<account>ACC-\d{4})$', (ref_str or '').strip())
    return {"category": m.group("category"), "account": m.group("account")} if m else None

with_prefix = validate_categorized_account("CORP-ACC-1032")
bare = validate_categorized_account("ACC-1032")
print(f"[ACC-1032] With prefix: {with_prefix} | Bare: {bare}")
```

### Q33. Extract currency and amount for banking case 1; use sample reference `ACC-1033` and explain the result.

**Hints**

1. Write representative valid and invalid samples before writing the pattern.
2. Use raw strings, named groups for extraction, and `fullmatch` for whole-field validation.
3. Never treat regex alone as authorization or full semantic validation; verify dates and checksums separately.

**Edge cases to test:** empty input, invalid types, boundary values, and a realistic failure path.

```python
import re

def extract_currency_and_amount(text):
    pattern = r'(?P<curr>USD|EUR|GBP|JPY|[\$€£¥])\s*(?P<amt>\d{1,3}(?:,\d{3})*(?:\.\d{2})?|\d+(?:\.\d{2})?)'
    results = []
    for m in re.finditer(pattern, text or ''):
        results.append((m.group("curr"), float(m.group("amt").replace(',', ''))))
    return results

memo = "Settlement for ACC-1033: debited USD 15,200.50 and credited €13,850.00."
print(f"[ACC-1033] Currency Pairs: {extract_currency_and_amount(memo)}")
```

### Q80. Create a regex-based data quality report for banking case 1; use sample reference `ACC-1080` and explain the result.

**Hints**

1. Write representative valid and invalid samples before writing the pattern.
2. Use raw strings, named groups for extraction, and `fullmatch` for whole-field validation.
3. Never treat regex alone as authorization or full semantic validation; verify dates and checksums separately.

**Edge cases to test:** empty input, invalid types, boundary values, and a realistic failure path.

```python
import re

VALIDATORS = {
    "account_id": re.compile(r'^ACC-\d{4}$'),
    "amount": re.compile(r'^\$\d+(?:\.\d{2})?$'),
    "email": re.compile(r'^[A-Za-z0-9._%+-]+@[A-Za-z0-9.-]+\.[A-Za-z]{2,}$'),
    "status": re.compile(r'^(?:APPROVED|PENDING|REJECTED)$')
}

def generate_data_quality_report(records):
    total = len(records or [])
    if total == 0:
        return {"total": 0, "valid": 0, "pass_rate": 0.0, "errors": {}}
    valid_count = 0
    errors = {f: 0 for f in VALIDATORS}
    for r in records:
        is_valid = True
        for f, pat in VALIDATORS.items():
            if not pat.fullmatch(str(r.get(f, '')).strip()):
                is_valid = False
                errors[f] += 1
        if is_valid:
            valid_count += 1
    return {"total": total, "valid": valid_count, "pass_rate": round((valid_count / total) * 100, 1), "errors": errors}

batch = [
    {"account_id": "ACC-1080", "amount": "$500.00", "email": "client@apexbank.com", "status": "APPROVED"},
    {"account_id": "ACC-2002", "amount": "$1,000", "email": "invalid-email", "status": "PENDING"},
    {"account_id": "ACC-3003", "amount": "$250.25", "email": "user@test.org", "status": "APPROVED"}
]
report = generate_data_quality_report(batch)
print(f"[ACC-1080] Quality Report: {report}")
```

```python
import re

def tokenize_messy_receipt(raw_text):
    token_spec = [
        ('REF',    r'ACC-\d{4}'),
        ('DATE',   r'\d{4}-\d{2}-\d{2}'),
        ('AMOUNT', r'\$\d+(?:\.\d{2})?'),
        ('TAX',    r'TAX:\d+%'),
        ('SKIP',   r'\s+'),
        ('MISC',   r'[^\s]+'),
    ]
    master = '|'.join(f'(?P<{name}>{pat})' for name, pat in token_spec)
    tokens = []
    for mo in re.finditer(master, raw_text or ''):
        kind, val = mo.lastgroup, mo.group()
        if kind != 'SKIP':
            tokens.append((kind, val))
    return tokens

receipt = "RECEIPT*** DATE 2026-05-20 REF ACC-1091 TOTAL $240.50 TAX:8% ***APPROVED"
tokens = tokenize_messy_receipt(receipt)
print(f"[ACC-1091] Stream: {tokens[:4]} ... Total tokens: {len(tokens)}")
```
