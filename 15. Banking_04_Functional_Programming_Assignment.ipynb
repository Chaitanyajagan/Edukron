### Q35. Write a decorator that logs checkout calls for banking case 1; use sample reference `ACC-1035` and explain the result.

```python
def log_checkout(func):
    def wrapper(*args, **kwargs):
        print(f"Logging checkout: {func.__name__}{args}")
        return func(*args, **kwargs)
    return wrapper

@log_checkout
def checkout(acc_id, amount):
    return f"Paid ${amount:.2f} from {acc_id}"

print(f"[ACC-1035] Result: {checkout('ACC-1035', 1450.75)}")
```

```python
from functools import reduce

transactions = [{"amt": 100.0, "valid": True}, {"amt": 50.0, "valid": False}, {"amt": 200.0, "valid": True}]

valid_tx = filter(lambda t: t["valid"], transactions)
with_bonus = map(lambda t: t["amt"] * 1.02, valid_tx)
total = reduce(lambda a, b: a + b, with_bonus, 0.0)

print(f"[ACC-1036] Total with Bonus: ${total:.2f}")
```
