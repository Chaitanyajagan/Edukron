```python
def calculate_one_pass_metrics(transactions):
    if not transactions:
        return {"count": 0, "total": 0.0, "min": 0.0, "max": 0.0, "avg": 0.0}
    total = count = 0
    min_val, max_val = float("inf"), float("-inf")
    for tx in transactions:
        val = float(tx.get("amount", 0.0))
        total += val
        count += 1
        min_val = min(min_val, val)
        max_val = max(max_val, val)
    return {"count": count, "total": round(total, 2), "min": round(min_val, 2), "max": round(max_val, 2), "avg": round(total/count, 2)}

txs = [{"amount": 120.0 * 1}, {"amount": 450.0 * 1}, {"amount": 230.0 * 1}]
metrics = calculate_one_pass_metrics(txs)
print(f"[ACC-1044] Metrics: Total=${metrics['total']:.2f}, Avg=${metrics['avg']:.2f}, Count={metrics['count']}")
```

### Q100. Design an end-to-end transactions analytics pipeline for banking case 3; use sample reference `ACC-1100` and explain the result.

**Hints**

1. First express the input and return value of one pure function.
2. Use `filter` for selection, `map` for transformation, and `reduce` only when a clearer built-in is not available.
3. Test the pipeline with an empty collection and avoid mutating account records in place.

**Edge cases to test:** empty input, invalid types, boundary values, and a realistic failure path.

```python
from functools import reduce

def analytics_pipeline(raw_transactions, high_risk_threshold=15000.0):
    cleaned = [t for t in raw_transactions if isinstance(t, dict) and t.get("amount", 0) > 0]
    
    def normalize(t):
        return {
            "ref": t.get("ref", "UNKNOWN"),
            "amount": float(t["amount"]),
            "is_flagged": float(t["amount"]) >= high_risk_threshold
        }
    normalized = list(map(normalize, cleaned))
    
    def aggregate(acc, t):
        acc["total_volume"] += t["amount"]
        acc["transaction_count"] += 1
        if t["is_flagged"]:
            acc["flagged_count"] += 1
        return acc

    summary = reduce(aggregate, normalized, {"total_volume": 0.0, "transaction_count": 0, "flagged_count": 0})
    summary["total_volume"] = round(summary["total_volume"], 2)
    return summary

sample_txs = [
    {"ref": "ACC-1100", "amount": 5400.0},
    {"ref": "ACC-1100", "amount": 24600.0},
    {"ref": "ACC-1100", "amount": 1350.0},
    {"ref": "INVALID", "amount": -100.0}
]
report = analytics_pipeline(sample_txs)
print(f"[ACC-1100] Pipeline Summary: Volume=${report['total_volume']:,.2f}, Count={report['transaction_count']}, Flagged={report['flagged_count']}")
```
